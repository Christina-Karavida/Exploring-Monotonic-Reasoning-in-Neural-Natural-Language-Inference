# Taxonomy Exploration


This notebook provides a review of the taxonomy we generated for the purposes of this thesis.

It serves two main goals:

1. to help the user navigate the hyponym-hypernym pairs  
2. to calculate the statistics of the taxonomy: i.e. the total number of categoris, hyponyms and hyponym–hypernym pairs


```python
from taxonomy import CATEGORY_CONFIG

from taxonomy_helpers import (category_counts,
    category_examples,
    hypo_hyper_pairs,
    total_pairs,
    hierarchy_breakdown
)
```

## 1. Category Counts

We begin by checking how many examples are available in each of the **44 available categories**: 

```python
category_counts(CATEGORY_CONFIG)
```

## 2. Examples per Category

Next, we inspect some samples from a given category:


```python
category_examples(CATEGORY_CONFIG, "Dogs", n=10)
category_examples(CATEGORY_CONFIG, "Toys", n=10)
category_examples(CATEGORY_CONFIG, "Shapes", n=10)
```

## 3. Examples of Hyponym–Hypernym Pairs

We then generate hyponym–hypernym pairs to illustrate how inference chains are built.  
For example, from the `Dogs` category, a sample entry such as `beagle` expands into multiple hypernyms:  
`beagle → dogs`, `beagle → canines`, `beagle → carnivores`, `beagle → mammals`.  

```python
hypo_hyper_pairs(CATEGORY_CONFIG, "Dogs", n=1) 
hypo_hyper_pairs(CATEGORY_CONFIG, "Viruses", n=1)
```

## 4. Total Pairs and Category Breakdown

We calculate the total number of hyponym–hypernym pairs per category:

```python
pair_counts = total_pairs(CATEGORY_CONFIG)
for cat, c in pair_counts.items():
    print(f"{cat:<15} -> {c}")
```

## 5. Hierarchy

A summry of the hierarchy, showing the number of examples/category, the number of hypernyms as well as some examples.


```python
hierarchy_breakdown(CATEGORY_CONFIG, n_examples=3)
```
